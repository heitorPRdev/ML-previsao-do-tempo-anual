```python
import pandas as pd
df = pd.read_csv("data.csv")
x = df[["temp_max","temp_min","vel_max","umidade","chance_chuva"]]
y = df[["content"]]
df
```

```python
from sklearn.model_selection import train_test_split
x_train,X_teste,y_train,Y_test = train_test_split(x,y,test_size=0.5,random_state=42)
X_teste
```

```python
from sklearn.tree import DecisionTreeClassifier
cfg = DecisionTreeClassifier()
cfg.fit(x_train,y_train)
pred = cfg.predict(X_teste)
cfg.score(X_teste,Y_test)
```

```python
import matplotlib.pyplot as plt
plt.hist(pred)
plt.show()
```
